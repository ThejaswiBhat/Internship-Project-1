# **Cohort Analysis**

**Creating Invoice Month**

```python
df['InvoiceMonth'] = df['InvoiceDate'].dt.to_period('M')
```

```python
df.head(1)
```

**Creating Cohort Month**

```python
cohort_month = (
    df.groupby('CustomerID')['InvoiceMonth']
      .min()
)
```

```python
df['CohortMonth'] = df['CustomerID'].map(cohort_month)
```

```python
df.head(1)
```

**Creating Cohort Index**

```python
invoice_year = df['InvoiceMonth'].dt.year
cohort_year = df['CohortMonth'].dt.year
```

```python
invoice_month = df['InvoiceMonth'].dt.month
cohort_month = df['CohortMonth'].dt.month
```

```python
df['CohortIndex'] = (
    (invoice_year - cohort_year) * 12
    + (invoice_month - cohort_month)
)
```

```python
df.head(4)
```

**Count Unique Customers**

```python
cohort_data = (
    df.groupby(['CohortMonth', 'CohortIndex'])['CustomerID']
      .nunique()
      .reset_index()
)
```

```python

cohort_data.head(10)
```

**Creating the Retention Matrix**

```python
retention_matrix = cohort_data.pivot(
    index='CohortMonth',
    columns='CohortIndex',
    values='CustomerID'
)
```

```python
retention_matrix
```

**Retention Percentage Matrix**

```python
cohort_size = retention_matrix[0]

retention_percentage = (
    retention_matrix.divide(cohort_size, axis=0)
    * 100
)
```

```python
retention_percentage.round(2)
```

**Cohort Retention Analysis**

- All cohorts begin with 100% retention during the acquisition month (Month 0).
- Retention drops sharply after the first month, indicating substantial early customer churn.
- Most cohorts retain approximately 10–25% of customers after one month.
- Some cohorts show higher retention in later months, suggesting that customers do not necessarily purchase every month.
- Long-term retention remains relatively low, highlighting opportunities for customer engagement and retention strategies.

**Heatmap Visualization of Retention Percentage**

```python
plt.figure(figsize=(14, 8))

sns.heatmap(
    retention_percentage,
    annot=True,
    fmt='.1f',
    cmap='Blues',
    linewidths=0.5
)

plt.title('Customer Retention Heatmap')
plt.xlabel('Months Since Acquisition')
plt.ylabel('Cohort Month')

plt.tight_layout()
plt.show()
```

# **Creating a Segment Summary To Calculate CLTV**

**The objective is to estimate Customer Lifetime Value (CLTV) by calculating customer spending behavior and purchase frequency. This helps identify high-value customers and supports data-driven business decisions.**

**Customer Value estimates how much revenue a customer generates.**

```python
country_cltv = df.groupby("Country").agg(
    Total_Revenue=("Revenue", "sum"),
    Total_Orders=("InvoiceNo", "nunique"),
    Total_Customers=("CustomerID", "nunique")
).reset_index()
```

```python
country_cltv.head()
```

**Calculate Average Order Value for each Country**

**Average Order Value tells us how much a customer spends on average each time they place an order.**

```python
country_cltv["Average_Order_Value"] = (
    country_cltv["Total_Revenue"] /
    country_cltv["Total_Orders"]
)

country_cltv.head()
```

**Calculating Purchase Frequency**

```python
country_cltv["Purchase_Frequency"] = (
    country_cltv["Total_Orders"] /
    country_cltv["Total_Customers"]
)

country_cltv.head()
```

**Calculating Historical CLTV**

* Historical Customer Lifetime Value (CLTV) estimates the average revenue generated by a customer using historical purchase data.

```python
country_cltv["Historical_CLTV"] = (
    country_cltv["Average_Order_Value"] *
    country_cltv["Purchase_Frequency"]
)

country_cltv.head()
```

```python
country_cltv = country_cltv.sort_values(
    by="Historical_CLTV",
    ascending=False
)

country_cltv.head(10)
```

```python
top10_cltv = country_cltv.head(10)

plt.figure(figsize=(10,6))
plt.bar(top10_cltv["Country"], top10_cltv["Historical_CLTV"])

plt.title("Top 10 Countries by Historical CLTV")
plt.xlabel("Country")
plt.ylabel("Historical CLTV")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

## Key Findings

- EIRE recorded the highest Historical CLTV (£3053.60), mainly due to a very high purchase frequency.
- The Netherlands ranked second, with customers placing frequent orders and having a high average order value.
- Australia had the highest average order value among the leading countries, indicating higher spending per transaction.
- France generated high total revenue but had a comparatively lower Historical CLTV because of its lower average order value.
